```python
#import libraries
import numpy as np 
import pandas as pd 
```

```python
#loading excel file
df = pd.read_excel("sales_dataset.xlsx")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
#making a copy of the original dataset
raw_df = df.copy()
```

```python
#standardize column names
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)

print(df.columns.tolist())
```

```python
#Check dataset information
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.info()
```

```python
#Check duplicate rows
duplicate_count = df.duplicated().sum()

print("Exact duplicate rows:", duplicate_count)
```

```python
#drop duplicate
df = df.drop_duplicates()

print("Shape after removing exact duplicates:", df.shape)
```

```python
#repeated and duplicate orders 
duplicate_order_ids = df["order_id"].duplicated().sum()

print("Repeated Order ID rows:", duplicate_order_ids)
print("Unique Orders:", df["order_id"].nunique())
```

```python
#droping unnessary columns 
columns_to_drop = [
    "unnamed:_22",
    "currency",
    "ship_country",
    "fulfilled_by"
]

df = df.drop(
    columns=columns_to_drop,
    errors="ignore"
)

print("Columns after removal:")
print(df.columns.tolist())
```

```python
# first 10 rows
df.head(10)
```

```python
#Remove unnecessary spaces
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
#converting text to lowercase
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].str.lower()
```

```python
#Clean city names
city_mapping = {
    "bengalore": "bangalore",
    "banglore": "bangalore",
    "bengaluru": "bangalore",
    "new delhi": "new delhi",
    "delhi": "delhi"
}

df["ship_city"] = df["ship_city"].replace(city_mapping)
```

```python
print(df["ship_city"].value_counts().head(20))
```

```python
#cleaning state names
state_mapping = {
    "bihar": "bihar",
    "bhihar": "bihar",

    "orissa": "odisha",
    "orissa ": "odisha",

    "rajsthan": "rajasthan",
    "rajshthan": "rajasthan",

    "rj": "rajasthan",

    "pb": "punjab",

    "mp": "madhya pradesh",

    "up": "uttar pradesh",

    "uk": "uttarakhand",

    "tn": "tamil nadu",

    "ka": "karnataka",

    "kl": "kerala",

    "ap": "andhra pradesh",

    "ts": "telangana",

    "wb": "west bengal",

    "mh": "maharashtra",

    "gj": "gujarat",

    "hr": "haryana",

    "dl": "delhi",
     "nl": "nagaland",
    "ar": "arunachal pradesh",
    "punjab/mohali/zirakpur": "punjab",
    "pondicherry": "puducherry"
}

df["ship_state"] = df["ship_state"].replace(state_mapping)
```

```python
print(df["ship_state"].value_counts())
```

```python
df = df[df["ship_state"] != "apo"]
```

```python
for col in df.select_dtypes(include="object").columns:
    df[col] = (
        df[col]
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
    )
```

```python
print("Rows after removing apo:", len(df))
print(df["ship_state"].value_counts().tail(10))
```

```python
#date
df["date"] = pd.to_datetime(
    df["date"],
    errors="coerce",
    dayfirst=True
)
```

```python
print(df["date"].dtype)
```

```python
#Convert numerical columns
df["qty"] = pd.to_numeric(
    df["qty"],
    errors="coerce"
)

df["amount"] = pd.to_numeric(
    df["amount"],
    errors="coerce"
)
```

```python
print(df[["qty", "amount"]].dtypes)
```

```python
#checking missing values
missing = df.isnull().sum()

print(missing[missing > 0])
```

```python
missing_report = pd.DataFrame({
    "missing_count": df.isnull().sum(),
    "missing_percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

missing_report = missing_report[
    missing_report["missing_count"] > 0
].sort_values(
    "missing_percentage",
    ascending=False
)

print(missing_report)
```

```python
#Handle missing categorical values
categorical_columns = [
    "status",
    "fulfilment",
    "sales_channel",
    "ship_service_level",
    "category",
    "size",
    "courier_status",
    "ship_city",
    "ship_state",
    "promotion_ids",
    "b2b"
]

for col in categorical_columns:
    if col in df.columns:
        df[col] = df[col].fillna("unknown")
```

```python
#handling missing postal code
df["ship_postal_code"] = (
    pd.to_numeric(
        df["ship_postal_code"],
        errors="coerce"
    )
    .astype("Int64")
    .astype("string")
)
```

```python
df["ship_postal_code"] = df["ship_postal_code"].fillna("unknown")
```

```python
#Check impossible values(negative ammount and quantaties)
print(
    "Negative quantities:",
    (df["qty"] < 0).sum()
)
print(
    "Negative amounts:",
    (df["amount"] < 0).sum()
)
print(
    "Zero quantity rows:",
    (df["qty"] == 0).sum()
)
print(
    "Missing/invalid dates:",
    df["date"].isna().sum()
)
```

```python
print("Status:")
print(df["status"].unique())

print("\nFulfilment:")
print(df["fulfilment"].unique())

print("\nCategory:")
print(df["category"].unique())

print("\nSales Channel:")
print(df["sales_channel"].unique())

print("\nShip Service Level:")
print(df["ship_service_level"].unique())
```

```python
#Check the cleaned dataset
print("Final shape:", df.shape)

df.head()
```

```python
df.info()
```

```python
#Final missing-value check
final_missing = pd.DataFrame({
    "missing_count": df.isnull().sum(),
    "missing_percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

final_missing = final_missing[
    final_missing["missing_count"] > 0
]

print(final_missing)
```

```python
#Final duplicate check
print(
    "Exact duplicate rows:",
    df.duplicated().sum()
)

print(
    "Unique orders:",
    df["order_id"].nunique()
)

print(
    "Total rows:",
    len(df)
)
```

```python
#Number of unique values in each columns
unique_report = pd.DataFrame({
    "column": df.columns,
    "unique_values": [
        df[col].nunique()
        for col in df.columns
    ]
})

print(unique_report)
```

```python
#total units 
total_revenue = df["amount"].sum()

print("Total Revenue: ₹", round(total_revenue, 2))
```

```python
order_revenue = (
    df.groupby("order_id")["amount"]
    .sum()
)

aov = order_revenue.mean()

print("Average Order Value: ₹", round(aov, 2))
```

```python
cancelled_orders = (
    df.loc[
        df["status"] == "cancelled",
        "order_id"
    ].nunique()
)

cancellation_rate = (
    cancelled_orders / total_orders * 100
)

print(
    "Cancellation Rate:",
    round(cancellation_rate, 2),
    "%"
)
```
